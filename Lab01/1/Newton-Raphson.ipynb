```python
import numpy as np

def newton_raphson(grad_f, hessian_f, x0, tol = 1e-12, max_inter = 100):
    x_n = np.array(x0, dtype = float)

    for i in range(max_inter):
        grad = grad_f(x_n)
        hess = hessian_f(x_n)

        if np.linalg.norm(grad) < tol:
            return x_n

        try :
            delta_x = np.linalg.solve(hess, -grad)
        except np.linalg.LinAlgError:
            return None

        x_n = x_n + delta_x

    return x_n
```
